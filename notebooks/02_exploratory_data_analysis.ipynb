```python
import sys
from pathlib import Path

project_root = Path.cwd().parent
if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

import pandas as pd
import numpy as np
import plotly.io as pio
pio.renderers.default = "iframe_connected+jupyterlab+colab"

from src.config import PROCESSED_DATA_PATH, TARGET_COL
from src.analysis.stats import CreditRiskStatsAnalyzer
from src.analysis.financial import FinancialBehaviorAnalyzer
from src.visualization.plots import CreditRiskVisualizer

df = pd.read_csv(PROCESSED_DATA_PATH)
print(f"Loaded processed portfolio: {df.shape[0]:,} rows x {df.shape[1]} columns")
```

## 1. Target Variable Analysis ()
Let us inspect the distribution of default outcomes across the portfolio.

```python
stats_engine = CreditRiskStatsAnalyzer(df)
target_dist = stats_engine.get_target_distribution()
for k, v in target_dist.items():
    print(f"{k}: {v}")

viz = CreditRiskVisualizer(df)
fig_target = viz.plot_target_distribution()
fig_target.show()
```

## 2. Advanced Summary Statistics of Numerical Features

```python
summary_stats = stats_engine.get_summary_statistics()
summary_stats.round(2)
```

## 3. Monotonic Risk Gradient: Default Rate by Loan Grade
Credit risk assessment requires that higher credit risk grades (from A to G) exhibit monotonically increasing default rates.

```python
grade_breakdown = stats_engine.default_rate_by_feature("loan_grade")
print(grade_breakdown.to_string(index=False))

fig_grade = viz.plot_loan_grade_default_analysis()
fig_grade.show()
```

## 4. Multi-Dimensional Risk Heatmap: Loan Intent vs Home Ownership

```python
fig_heat = viz.plot_risk_heatmap("loan_grade", "loan_intent")
fig_heat.show()
```

## 5. Correlation Analysis & Feature Associations

```python
corr_with_target = stats_engine.compute_correlations()
print("Pearson correlation with target (loan_status):")
print(corr_with_target)

fig_corr = viz.plot_correlation_matrix()
fig_corr.show()
```

## 6. Financial Behavior & Debt Burden Inflection Point
Analyzing how borrower debt burden () influences default rates.

```python
fin_engine = FinancialBehaviorAnalyzer(df)
exposure_summary = fin_engine.portfolio_exposure()
print(exposure_summary.summary())

debt_analysis = fin_engine.debt_burden_analysis()
print("
Debt Burden Risk Distribution:")
print(debt_analysis.to_string(index=False))

fig_scatter = viz.plot_financial_behavior_scatter()
fig_scatter.show()
```

## 7. Weight of Evidence (WoE) and Information Value (IV) Ranking
Information Value quantifies the predictive separation capacity of each feature without assuming linearity.

```python
iv_ranking = stats_engine.rank_all_features_by_iv()
print(iv_ranking.to_string(index=False))

fig_iv = viz.plot_iv_ranking(iv_ranking)
fig_iv.show()
```

Compare borrower distributions for Loan Amount, Debt Burden, and Annual Income.
